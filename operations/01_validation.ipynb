```python
def crossed_sem(da, a="number", b="sample", include_resid = False):
    # Crossed standard error
    # Size of dims (number, sample)
    na, nb = da.sizes[a], da.sizes[b]
    # grand mean
    grand = da.mean((a, b))
    m_a = da.mean(b)                     # per-member means (over samples)
    m_b = da.mean(a)                     # per-sample means (over members)

    resid = da - m_a - m_b + grand       # interaction/residual term
    if include_resid:
        var_e = (resid**2).sum((a, b)) / ((na - 1) * (nb - 1))
    else:
        var_e = 0
    
    var_mean = (m_a.var(a, ddof=1) / na
                + m_b.var(b, ddof=1) / nb
                - var_e / (na * nb))
    return grand, np.sqrt(var_mean.clip(min=0))
```

```python
mean_sahel, se_sahel = crossed_sem(impact_sahel.sum(dim = ['month', 'ISO']), a="number", b="sample", include_resid = True)
```

```python
mean_seasia, se_seasia = crossed_sem(impact_seasia.sum(dim = ['month', 'ISO']), a="number", b="sample", include_resid = True)
```

```python
# Crossed standard error: Example: Single ensemble member
da = impact.sum(dim = 'month')
a = 'number'
b = 'sample'
# Size of dims (number, sample)
na, nb = da.sizes[a], da.sizes[b]

# pick the median member (by its mean over samples)
m_a = da.mean([d for d in da.dims if d != a]).compute()   # one mean per member, in memory
median_idx = int(abs(m_a - m_a.median(a)).argmin(a))
default_sample = 50

samples = da.isel({a: median_idx})        # (sample,)  – parameter spread, median member only
members = da.isel({b: default_sample})    # (number,)  – ensemble spread at default parameters

# central estimate: ensemble mean, shifted by the mean parameter effect
grand = members.mean(a) + (samples.mean(b) - samples.isel({b: default_sample}))

var_mean = (members.var(a, ddof=1) / na
            + samples.var(b, ddof=1) / nb)
se = np.sqrt(var_mean)
```
